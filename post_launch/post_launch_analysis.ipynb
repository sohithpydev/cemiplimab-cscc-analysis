# Cemiplimab (Libtayo®): Post-launch Evidence & Real-world Outcomes

This notebook performs reproducible analysis and visualization for Phase 7: Post-launch Evidence, Real-world Adoption & Long-term Outcomes of cemiplimab, specifically:
1. **Clinical Trial vs. Real-World Efficacy**: Grouped bar chart comparing Objective Response Rates (ORR) and Complete Response (CR) rates between EMPOWER-CSCC-1 and real-world cohorts (CASE, REACT-CEMI, Siena).
2. **Real-world Survival Benchmarks**: Bar chart of median PFS and OS (in months) reported across RWE studies.

## Setup and Imports

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set style for publication-quality figures
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'figure.titlesize': 16,
    'savefig.bbox': 'tight',
    'savefig.dpi': 300
})

# Create figures directory if it doesn't exist
os.makedirs('../figures', exist_ok=True)
print("Setup complete.")
```

## Part 1: Load RWE Summary Data

```python
rwe_df = pd.read_csv('post_launch_summary.csv')
rwe_df
```

## Part 2: Clinical Trial vs. Real-world Efficacy Comparison
We compare the Objective Response Rate (ORR) and Complete Response (CR) rate of the pivotal Phase 2 trial (EMPOWER-CSCC-1) against real-world cohorts.

```python
efficacy_data = {
    'Study / Cohort': [
        'EMPOWER-CSCC-1 (Trial)', 'EMPOWER-CSCC-1 (Trial)',
        'CASE Study (RWE)', 'CASE Study (RWE)',
        'REACT-CEMI (RWE)', 'REACT-CEMI (RWE)',
        'Siena Cohort (RWE)', 'Siena Cohort (RWE)'
    ],
    'Metric': ['ORR', 'CR', 'ORR', 'CR', 'ORR', 'CR', 'ORR', 'CR'],
    'Percentage': [47.2, 16.1, 43.7, 11.2, 42.0, 8.0, 66.6, 18.5]
}
eff_df = pd.DataFrame(efficacy_data)

fig, ax = plt.subplots(figsize=(9, 5.5))
sns.barplot(x='Study / Cohort', y='Percentage', hue='Metric', data=eff_df, palette=['#1f77b4', '#aec7e8'], edgecolor='black', ax=ax)

ax.set_ylabel('Response Rate (%)', fontweight='bold')
ax.set_xlabel('Study Cohort (Clinical Trial vs. Real-world Observational)', fontweight='bold')
ax.set_title('Efficacy Comparison: Clinical Trial vs. Real-World Outcomes', fontweight='bold', pad=15)
ax.set_ylim(0, 80)
ax.legend(title='Response Category')

# Annotate bar values
for p in ax.patches:
    height = p.get_height()
    if height >= 0:
        ax.annotate(f'{height:.1f}%',
                    xy=(p.get_x() + p.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.savefig('../figures/rwe_vs_clinical_trial.png', dpi=300)
plt.show()
```

## Part 3: Real-world Survival Benchmarks
We visualize the median Progression-Free Survival (PFS) and Overall Survival (OS) benchmarks (in months) reported across RWE studies.

```python
survival_data = {
    'Study': ['CASE Study', 'REACT-CEMI', 'REACT-CEMI', 'US RWD'],
    'Metric': ['Median PFS', 'Median PFS', 'Median OS', 'Median OS'],
    'Months': [14.7, 8.6, 21.0, 24.8]
}
surv_df = pd.DataFrame(survival_data)

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x='Study', y='Months', hue='Metric', data=surv_df, palette='muted', edgecolor='black', ax=ax)

ax.set_ylabel('Duration (Months)', fontweight='bold')
ax.set_xlabel('Real-World Observational Study', fontweight='bold')
ax.set_title('Real-World Survival Benchmarks for Cemiplimab in cSCC', fontweight='bold', pad=15)
ax.set_ylim(0, 30)
ax.legend(title='Survival Metric')

for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{height:.1f} mos',
                    xy=(p.get_x() + p.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.savefig('../figures/rwe_survival_benchmarks.png', dpi=300)
plt.show()
```
