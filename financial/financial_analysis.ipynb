## Part 2: Geographic Pricing Comparison
We convert local pricing to normalized USD values to evaluate pricing discrepancies. We use 2026 exchange rates:
*   1 GBP = 1.25 USD
*   1 AUD = 0.66 USD
*   1 USD = 150 JPY (1 JPY = 0.0067 USD)

```python
pricing_data = [
    {'Region': 'United States', 'Local Price': 11703.03, 'Currency': 'USD', 'USD Price': 11703.03, 'Pricing Type': 'WAC List Price'},
    {'Region': 'United Kingdom', 'Local Price': 4650.00, 'Currency': 'GBP', 'USD Price': 4650.00 * 1.25, 'Pricing Type': 'NHS List Price'},
    {'Region': 'Australia', 'Local Price': 7381.23, 'Currency': 'AUD', 'USD Price': 7381.23 * 0.66, 'Pricing Type': 'PBS Dispensed Price'},
    {'Region': 'Japan', 'Local Price': 450437.00, 'Currency': 'JPY', 'USD Price': 450437.00 * 0.0067, 'Pricing Type': 'NHI Price List'}
]
pricing_df = pd.DataFrame(pricing_data)

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x='Region', y='USD Price', data=pricing_df, palette='viridis', edgecolor='black', width=0.5, ax=ax)

ax.set_ylabel('Normalized Price per 350 mg Vial ($ USD)', fontweight='bold')
ax.set_xlabel('Geographic Region', fontweight='bold')
ax.set_title('Libtayo® (cemiplimab) Geographic Pricing Discrepancy (USD)', fontweight='bold', pad=15)
ax.set_ylim(0, 14000)

# Annotate pricing details on top of the bars
for i, p in enumerate(ax.patches):
    usd_val = pricing_df['USD Price'][i]
    local_val = pricing_df['Local Price'][i]
    curr = pricing_df['Currency'][i]
    p_type = pricing_df['Pricing Type'][i]
    
    ax.annotate(f'${usd_val:,.2f} USD\n({curr} {local_val:,.0f} list)',
                xy=(p.get_x() + p.get_width() / 2, usd_val),
                xytext=(0, 5),
                textcoords="offset points",
                ha='center', va='bottom', fontsize=9, fontweight='bold')
    
    ax.text(p.get_x() + p.get_width()/2, 500, p_type, ha='center', color='white', rotation=90, fontweight='bold')

plt.tight_layout()
plt.savefig('../figures/financial_pricing_comparison.png', dpi=300)
plt.show()
```

## Part 3: Annual Global Sales Trend

```python
sales_data = {
    'Year': ['2023', '2024'],
    'Sales': [868.8, 1217.0]  # Millions USD
}
sales_df = pd.DataFrame(sales_data)

fig, ax = plt.subplots(figsize=(6, 5))
sns.barplot(x='Year', y='Sales', data=sales_df, palette='rocket', edgecolor='black', width=0.4, ax=ax)

ax.set_ylabel('Global Net Product Sales ($ Millions USD)', fontweight='bold')
ax.set_xlabel('Reporting Year', fontweight='bold')
ax.set_title('Libtayo® Annual Global Sales Revenue\n(2023 vs 2024)', fontweight='bold', pad=15)
ax.set_ylim(0, 1500)

for i, p in enumerate(ax.patches):
    val = sales_df['Sales'][i]
    ax.annotate(f'${val:.1f}M',
                xy=(p.get_x() + p.get_width() / 2, val),
                xytext=(0, 5),
                textcoords="offset points",
                ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('../figures/financial_revenue_trend.png', dpi=300)
plt.show()
```
