2. **Geographic Coverage & Technology Categories**: Comparing cemiplimab patents across US, Europe, and Japan.

## Setup and Imports

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime

# Set style for publication-quality figures
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'figure.titlesize': 16,
    'savefig.bbox': 'tight',
    'savefig.dpi': 300
})

# Create figures directory if it doesn't exist
os.makedirs('../figures', exist_ok=True)
print("Setup complete.")
```

## Part 2: Exclusivity Timeline (Gantt Chart)
We visualize the timelines for regulatory data exclusivity (BPCIA in the US, EMA in the EU) and key US patents (composition, method of use, formulation).

```python
fig, ax = plt.subplots(figsize=(11, 5.5))

# Define elements and their start/end years
elements = [
    {'label': 'US FDA Regulatory Exclusivity (BPCIA)', 'start': 2018, 'end': 2030, 'color': '#aec7e8', 'type': 'Regulatory'},
    {'label': 'EU EMA Regulatory Exclusivity', 'start': 2019, 'end': 2029, 'color': '#ffbb78', 'type': 'Regulatory'},
    {'label': 'US Patent 10,737,113 (Composition)', 'start': 2020, 'end': 2035, 'color': '#1f77b4', 'type': 'Patent'},
    {'label': 'US Patent 9,987,500 (Composition)', 'start': 2018, 'end': 2035, 'color': '#1f77b4', 'type': 'Patent'},
    {'label': 'US Patent 10,457,725 (Method of Use)', 'start': 2019, 'end': 2037, 'color': '#2ca02c', 'type': 'Patent'},
    {'label': 'US Patent 11,603,407 (Formulation)', 'start': 2023, 'end': 2038, 'color': '#ff7f0e', 'type': 'Patent'}
]

y_labels = [el['label'] for el in elements]
y_pos = np.arange(len(elements))

for i, el in enumerate(elements):
    ax.barh(i, el['end'] - el['start'], left=el['start'], color=el['color'], edgecolor='black', height=0.5)
    # Label start and end dates
    ax.text(el['start'] - 0.3, i, str(el['start']), va='center', ha='right', fontsize=9, fontweight='semibold')
    ax.text(el['end'] + 0.3, i, str(el['end']), va='center', ha='left', fontsize=9, fontweight='bold')

ax.set_yticks(y_pos)
ax.set_yticklabels(y_labels, fontweight='bold')
ax.invert_yaxis()  # top-down view
ax.set_xlabel('Timeline (Years)', fontweight='bold')
ax.set_title('Cemiplimab (Libtayo®): US & EU Regulatory Exclusivity and Patent Expirations', fontweight='bold', pad=15)
ax.set_xlim(2014, 2042)

# Add vertical line for today (approx 2026)
ax.axvline(2026.5, color='red', linestyle='--', linewidth=1.5)
ax.text(2026.5, -0.7, 'Current Year\n(2026)', color='red', ha='center', va='bottom', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.savefig('../figures/patent_exclusivity_timeline.png', dpi=300)
plt.show()
```

## Part 3: Geographic Coverage & Patent Categories
We compare the cemiplimab patent portfolios across the United States, Europe, and Japan.

```python
# Construct a comparison table for active patents
data_geo = {
    'Jurisdiction': ['United States', 'United States', 'United States', 'Europe', 'Europe', 'Japan'],
    'Category': ['Composition of Matter', 'Method of Use', 'Formulation', 'Composition of Matter', 'Method of Use', 'Composition of Matter'],
    'Patents Count': [2, 1, 1, 1, 1, 1]
}

geo_df = pd.DataFrame(data_geo)

fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x='Jurisdiction', y='Patents Count', hue='Category', data=geo_df, palette='Blues_r', edgecolor='black', ax=ax)

ax.set_ylabel('Number of Active Patents', fontweight='bold')
ax.set_xlabel('Jurisdiction / Patent Office', fontweight='bold')
ax.set_title('Cemiplimab Active Patents by Category and Jurisdiction', fontweight='bold', pad=15)
ax.set_ylim(0, 3)
ax.legend(title='Technology Category')

# Annotate bar values
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{int(height)}',
                    xy=(p.get_x() + p.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('../figures/patent_jurisdiction_comparison.png', dpi=300)
plt.show()
```
