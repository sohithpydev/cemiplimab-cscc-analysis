# Cemiplimab (Libtayo®): Commercial Landscape & Market Positioning

This notebook performs reproducible analysis and visualization for Phase 5: Commercial Landscape & Market Positioning of cemiplimab, specifically:
1. **Libtayo Global Sales Growth**: Visualizing the net product sales revenue for 2023 and 2024.
2. **Competitor Efficacy Profile**: Grouped bar chart comparing Objective Response Rates (ORR) and Complete Response (CR) rates of key systemic therapies for advanced cSCC.

## Setup and Imports

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set style for publication-quality figures
sns.set_theme(style="whitegrid")
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.titlesize': 14,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'figure.titlesize': 16,
    'savefig.bbox': 'tight',
    'savefig.dpi': 300
})

# Create figures directory if it doesn't exist
os.makedirs('../figures', exist_ok=True)
print("Setup complete.")
```

## Part 1: Load Competitor Data

```python
comp_df = pd.read_csv('commercial_summary.csv')
comp_df
```

## Part 2: Libtayo Global Sales Growth
We visualize the net product sales revenue for 2023 and 2024 as disclosed in Regeneron's financial statements ($868.8 million in 2023 and $1,217.0 million in 2024, representing 40.1% year-on-year growth).

```python
sales_data = {
    'Year': ['2023', '2024'],
    'Sales': [868.8, 1217.0]  # in millions USD
}
sales_df = pd.DataFrame(sales_data)

fig, ax = plt.subplots(figsize=(6, 5))
sns.barplot(x='Year', y='Sales', data=sales_df, palette='Blues', edgecolor='black', width=0.5, ax=ax)

ax.set_ylabel('Global Net Product Sales ($ Millions USD)', fontweight='bold')
ax.set_xlabel('Reporting Year', fontweight='bold')
ax.set_title('Libtayo® (cemiplimab) Global Sales Performance\n(2023 vs 2024)', fontweight='bold', pad=15)
ax.set_ylim(0, 1500)

# Annotate percentages and values
for i, p in enumerate(ax.patches):
    val = sales_df['Sales'][i]
    ax.annotate(f'${val:.1f}M',
                xy=(p.get_x() + p.get_width() / 2, val),
                xytext=(0, 5),
                textcoords="offset points",
                ha='center', va='bottom', fontweight='bold')

# Annotate growth rate
ax.annotate('+40.1% YoY Growth',
            xy=(0.5, 1000), xytext=(0.5, 1300),
            arrowprops=dict(facecolor='black', shrink=0.08, width=1.5, headwidth=6),
            ha='center', color='green', fontweight='bold')

plt.tight_layout()
plt.savefig('../figures/commercial_revenue_growth.png', dpi=300)
plt.show()
```

## Part 3: Efficacy Comparison across Systemic Options
We compare the clinical response rates (Objective Response Rate [ORR] and Complete Response [CR]) of the main systemic options in advanced cSCC.

```python
efficacy_data = {
    'Therapy': ['Cemiplimab', 'Cemiplimab', 'Pembrolizumab', 'Pembrolizumab', 'Cetuximab', 'Cetuximab', 'Chemotherapy', 'Chemotherapy'],
    'Metric': ['ORR', 'CR', 'ORR', 'CR', 'ORR', 'CR', 'ORR', 'CR'],
    'Percentage': [47.2, 16.1, 42.6, 6.0, 28.0, 0.0, 34.0, 2.0]  # Standardized clinical cross-trial percentages
}

eff_df = pd.DataFrame(efficacy_data)

fig, ax = plt.subplots(figsize=(9, 5.5))
sns.barplot(x='Therapy', y='Percentage', hue='Metric', data=eff_df, palette=['#1f77b4', '#aec7e8'], edgecolor='black', ax=ax)

ax.set_ylabel('Response Rate (%)', fontweight='bold')
ax.set_xlabel('Systemic Treatment Regimen', fontweight='bold')
ax.set_title('Efficacy Comparison of Systemic Options in Advanced cSCC', fontweight='bold', pad=15)
ax.set_ylim(0, 60)
ax.legend(title='Response Category')

# Annotate bar values
for p in ax.patches:
    height = p.get_height()
    if height >= 0:
        ax.annotate(f'{height:.1f}%',
                    xy=(p.get_x() + p.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.savefig('../figures/commercial_efficacy_comparison.png', dpi=300)
plt.show()
```
